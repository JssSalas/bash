```python
rural  = df.loc[df["DM4"] == 2, "DM6"]
urbano = df.loc[df["DM4"] == 1, "DM6"]
n1, n2, N = len(rural), len(urbano), len(rural) + len(urbano)
print(f"n rural={n1}, n urbano={n2}")
print(f"N={N}")
```

```python
#Prueba directa
res = mannwhitneyu(rural, urbano, alternative="two-sided", method="asymptotic")
U1=res.statistic
U2 = n1 * n2 - U1
U = min(U1, U2)
```

```python
# z con corrección por empates
t = df["DM6"].value_counts().values
sigma = np.sqrt(n1 * n2 / 12 * ((N + 1) - (t**3 - t).sum() / (N * (N - 1))))
z = (U - n1 * n2 / 2) / sigma
r = abs(z) / np.sqrt(N)
```

```python
print(f"n rural={n1}, n urbano={n2}")
print(f"U_rural={U1:.1f}, U_urbano={U2:.1f}, suma={U1+U2:.0f} (debe ser {n1*n2})")
print(f"U={U:.1f}, z={z:.3f}, p={res.pvalue:.4f}, r={r:.3f}")
print("Medianas:", rural.median(), "(rural) vs", urbano.median(), "(urbano)")
print(pd.crosstab(df["DM4"], df["DM6"]))
```
